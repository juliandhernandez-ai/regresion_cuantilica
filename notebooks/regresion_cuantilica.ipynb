# Regresión lineal clásica

La regresión lineal estudia la relación entre una variable respuesta
$Y$ y una o más covariables $X$.

<br>

$$
Y = X\beta + \varepsilon
$$

<br>

Su objetivo principal es modelar la **media condicional**:

$$
\boxed{
E(Y\mid X=x)
}
$$

<br>

## Regresión cuantilica como alternativa

- El gráfico nos dice que la distribución de los salarios **cambia de forma** (se ensancha y se sesga) a medida que las empresas crecen.
- La **regresión lineal clásica** solo nos da una línea recta que representa el promedio, la cual es **engañosa**.
- La **regresión cuantílica** nos permite dibujar varias líneas (una para cada cuantil) que capturan la **verdadera forma cambiante** de los datos.

# Entonces, ¿qué podemos hacer?

En lugar de estudiar únicamente la media:

$$
E(Y\mid X=x)
$$

podemos estudiar diferentes puntos de la distribución condicional:

$$
\boxed{
Q_Y(\tau\mid X=x)
}
$$

donde

$$
\tau\in(0,1)
$$

representa el cuantil de interés.

<br>

Por ejemplo:

$$
\begin{array}{ccccc}
\tau=0.10 &
\tau=0.25 &
\tau=0.50 &
\tau=0.75 &
\tau=0.90
\end{array}
$$

<br>

$$
\text{10\%}
\qquad
\text{25\%}
\qquad
\textbf{Mediana}
\qquad
\text{75\%}
\qquad
\text{90\%}
$$

### Definición formal de la función cuantil

El cuantil de nivel $\tau$ de una variable aleatoria $Y$ se define formalmente mediante la **inversa generalizada** de la función de distribución:

$$
Q_Y(\tau)=\inf\left\{y:F_Y(y)\geq\tau\right\}, \qquad 0<\tau<1.
$$

>**Sin embargo, la documentación trabaja directamente con esta notación simplificada:**

$$Q_Y(\tau) = F_Y^{-1}(\tau)$$

# Regresión cuantílica
La regresión cuantílica es una técnica que permite estudiar cómo unas variables explicativas se relacionan con diferentes posiciones de la distribución de una variable respuesta, en lugar de estudiar solamente su promedio.

La regresión cuantílica permite modelar directamente
los cuantiles condicionales de $Y$.


<br>

$$
\boxed{
Q_Y(\tau\mid X=x)
=
\beta_0(\tau)
+
\beta_1(\tau)x_1
+\cdots+
\beta_p(\tau)x_p
}
$$

<br>

Una característica fundamental es que los coeficientes
pueden cambiar dependiendo del cuantil:

$$
\beta_j(0.10)
\neq
\beta_j(0.50)
\neq
\beta_j(0.90)
$$

<br>

$$
\boxed{
\text{La relación entre }X\text{ y }Y
\text{ puede ser diferente en distintas partes de la distribución.}
}
$$

**Recordemos**

 **En regresión lineal clásica** preguntamos:

> *“¿Cómo cambia, en promedio, $Y$ cuando cambia $X$?”*

En **regresión cuantílica** podemos hacer preguntas como:

* **Parte baja:** *“¿Cómo cambia $Y$ entre quienes están en la parte baja de la distribución?”*
* **Parte central:** *“¿Cómo cambia en la parte central?”*
* **Parte alta:** *“¿Y qué ocurre en la parte alta?”*

# La Función de Pérdida Cuantílica 



$$
\rho_\tau(u_i) = \begin{cases} \tau \cdot u_i & \text{si } u_i \ge 0 \quad \text{(subestimación)} \\ (\tau - 1) \cdot u_i & \text{si } u_i < 0 \quad \text{(sobreestimación)} \end{cases}
$$



### **¿Qué hace y para qué sirve en el modelo?**

* **Es la regla del juego:** No calcula los parámetros directamente, sino que define el "costo" o castigo por cada error cometido.
* **Penaliza con asimetría:** Aplica un peso desigual a los residuos positivos y negativos según el cuantil $\tau$ elegido.
* **Actúa como brújula:** Guía al algoritmo de optimización lineal para ajustar los coeficientes ($\beta$) hasta que la pérdida total sea la mínima posible.
* **Sustituye a MCO:** Al cambiar los errores al cuadrado por distancias absolutas ponderadas, desplaza la recta del promedio hacia la trayectoria del cuantil de interés.

### Teorema de Optimalidad del Cuantil (o Teorema de Elicitabilidad)

Dada una variable aleatoria continua $Y$ con función de distribución acumulada $F_Y(y)$ y una función de pérdida cuantílica (función *pinball*) definida por:

$$
\rho_{\tau}(u) = u \big(\tau - \mathbb{I}_{\{u < 0\}}\big)
$$

El valor constante $q^{*}$ que minimiza el valor esperado de la pérdida, es decir:

$$
q^{*} = \arg \min_{q} \, \mathbb{E}[\rho_{\tau}(Y - q)]
$$

existe, es único y coincide exactamente con el cuantil de nivel $\tau$ de la variable aleatoria $Y$:

$$
q^{*} = F_{Y}^{-1}(\tau)
$$


---
### Demostración 

Sea $Y$ una variable aleatoria continua con función de distribución acumulada $F_Y(y) = P(Y \le y)$ y función de densidad de probabilidad $f_Y(y)$. Queremos encontrar el estimador $q$ que minimiza la pérdida esperada:
$$\min_{q} \, \mathbb{E}[\rho_\tau(Y - q)]$$
---





Desarrollando la esperanza matemática según la definición por tramos de $\rho_\tau(u)$:

$$\mathbb{E}[\rho_\tau(Y - q)] = \int_{-\infty}^{\infty} \rho_\tau(y - q) f_Y(y) \, dy$$

$$\mathbb{E}[\rho_\tau(Y - q)] = \int_{-\infty}^{q} (\tau - 1)(y - q) f_Y(y) \, dy + \int_{q}^{\infty} \tau (y - q) f_Y(y) \, dy$$

Para hallar el mínimo, aplicamos la **Regla de Leibniz** para derivar bajo el signo de la integral con respecto a $q$, e igualamos a cero:

$$\frac{d}{dq} \mathbb{E}[\rho_\tau(Y - q)] = \int_{-\infty}^{q} (1 - \tau) f_Y(y) \, dy + \int_{q}^{\infty} (-\tau) f_Y(y) \, dy = 0$$

Separando las integrales:

$$(1 - \tau) \int_{-\infty}^{q} f_Y(y) \, dy - \tau \int_{q}^{\infty} f_Y(y) \, dy = 0$$

Sabiendo que $\int_{-\infty}^{q} f_Y(y) \, dy = F_Y(q)$ y que $\int_{q}^{\infty} f_Y(y) \, dy = 1 - F_Y(q)$, sustituimos:

$$(1 - \tau) F_Y(q) - \tau (1 - F_Y(q)) = 0$$

$$F_Y(q) - \tau F_Y(q) - \tau + \tau F_Y(q) = 0$$

$$F_Y(q) = \tau$$

Despejando $q$, obtenemos que la solución óptima $q^*$ es la función inversa de la distribución evaluada en $\tau$:

$$q^* = F_Y^{-1}(\tau)$$

### Conclusión Clave: El Puente hacia la Regresión


Queda demostrado formalmente que **el optimizador $q^*$ que minimiza la función de pérdida esperada es exactamente el cuantil $\tau$-ésimo** de la distribución de la variable aleatoria $Y$. 

El **Teorema de Optimalidad** transforma un concepto puramente estadístico en un problema de optimización:

* **Nueva Perspectiva:** Demuestra que un cuantil no solo se obtiene ordenando observaciones, sino minimizando una función de pérdida asimétrica (función *pinball*).

* **Propósito:** Es el pilar matemático que nos permite modelar de forma rigurosa la parte baja, central o alta de cualquier distribución de datos.
---

### 🎯 La Función Objetivo en Regresión Cuantílica

#### **1. Definición**
La **función objetivo** es una ecuación matemática de costo que calcula la **pérdida total acumulada** de un modelo para un nivel cuantílico específico ($\tau$). A diferencia de la regresión clásica, que eleva los errores al cuadrado, esta función utiliza la **función check (o pérdida pinball)** para dividir los residuos en dos grupos y penalizarlos de manera asimétrica:

$$
Q(\boldsymbol{\beta}_\tau) = \tau \sum_{y_i \ge \mathbf{x}_i^T \boldsymbol{\beta}_\tau} |y_i - \mathbf{x}_i^T \boldsymbol{\beta}_\tau| + (1 - \tau) \sum_{y_i < \mathbf{x}_i^T \boldsymbol{\beta}_\tau} |y_i - \mathbf{x}_i^T \boldsymbol{\beta}_\tau|
$$

*   **Si el modelo subestima** ($y_i \ge \hat{y}_i$), el error absoluto se multiplica por $\tau$.
*   **Si el modelo sobreestima** ($y_i < \hat{y}_i$), el error absoluto se multiplica por $(1 - \tau)$.

y buscamos los coeficientes que minimizan
la pérdida cuantílica total:

$$
\boxed{
\hat{\beta}(\tau)
=
\arg\min_{\beta\in\mathbb{R}^{p}}
\sum_{i=1}^{n}
\rho_\tau
\left(
y_i-x_i^\top\beta
\right)
}
$$

---

#### **2. ¿Por qué es importante?**

*   **Actúa como una "Balanza Probabilística":** Es el motor matemático que desplaza la recta de regresión verticalmente por todo el conjunto de datos. Al cambiar el valor de $\tau$, la función altera automáticamente los pesos de penalización, obligando al estimador a detenerse **exactamente** en el punto donde la masa de datos se divide en las proporciones deseadas (por ejemplo, 10% abajo y 90% arriba si $\tau = 0.10$).
*   **Permite mapear toda la distribución:** Al modificar de manera iterativa el parámetro $\tau$ en la función objetivo, se pueden estimar infinitas rectas de regresión. Esto te permite responder qué ocurre en la **parte baja, central o alta** de la variable respuesta, ofreciendo una radiografía completa del fenómeno en lugar de un único promedio.
*   **Aporta robustez extrema ante Outliers:** Al basarse en la suma de **desviaciones absolutas** ($|u|$) en lugar de errores al cuadrado ($u^2$), las observaciones atípicas o los valores extremos de la muestra no distorsionan la estimación. La recta no se ve atraída de forma desproporcionada por los valores atípicos, manteniendo la estabilidad del modelo.
*   **Habilita el uso de Programación Lineal:** Aunque sufre la desventaja de no ser diferenciable en cero (lo que impide usar derivadas simples), su estructura lineal por tramos permite transformarla en un problema de **Programación Lineal**. Gracias a esto, algoritmos exactos como el método *Simplex* o los métodos de *Punto Interior* pueden hallar la solución óptima global rápidamente y sin necesidad de aproximaciones.

# Selección de covariables en regresión cuantílica

En una regresión cuantílica multivariable se plantea:

$$
Q_Y(\tau\mid X)
=
\beta_0(\tau)+\beta_1(\tau)X_1+\cdots+\beta_p(\tau)X_p
$$

A diferencia de la regresión lineal clásica, el interés no se centra únicamente en explicar la **media** de $Y$, sino en estudiar diferentes puntos de su distribución.

### ¿Cómo seleccionar las covariables?

* **Pruebas de significancia:**
  Para cada cuantil $\tau$, se puede evaluar:

  $$
  H_0:\beta_j(\tau)=0
  \qquad\text{vs.}\qquad
  H_1:\beta_j(\tau)\neq0
  $$

* **Comparación de modelos:**
  Se pueden comparar modelos con diferentes conjuntos de covariables mediante criterios de ajuste.

* **Validación cruzada:**
  Permite evaluar el desempeño predictivo utilizando la pérdida cuantílica o **función check**.

> **Importante: la matriz de correlación puede utilizarse como herramienta exploratoria, pero no debería ser el único criterio de selección.**

### Una diferencia fundamental

**Regresión clásica**

$$
\text{Covariables}
\longrightarrow
\text{efecto sobre la media}
$$

**Regresión cuantílica**

$$
\text{Covariables}
\longrightarrow
\text{efecto sobre diferentes cuantiles}
$$




**Importante:**

| Variable | $\tau = 0.25$ | $\tau = 0.50$ | $\tau = 0.75$ |
| :--- | :--- | :--- | :--- |
| **$X_1$** | significativa | significativa | significativa |
| **$X_2$** | no significativa | significativa | significativa |
| **$X_3$** | significativa | no significativa | no significativa |

Esto es muy interesante porque una variable puede ser relevante para unos sectores de la distribución y no para otros.

Consideremos:

$$
Q_Y(\tau\mid X)
=
\beta_0(\tau)+\beta_1(\tau)X
$$

Entonces:

$$
\boxed{
\beta_1(\tau)

$$

representa el cambio en el cuantil $\tau$ de $Y$

<br>

Por ejemplo:

$$

# ¿Qué supuestos evitamos imponer?

Una regresión cuantílica no necesita asumir, para definir el cuantil condicional, una distribución paramétrica específica como la normalidad:

$$Y \mid X \sim \mathcal{N}(\mu, \sigma^2)$$

En lugar de especificar una distribución completa para los errores o la variable dependiente, modelamos directamente la estructura funcional del cuantil de interés:

$$\boxed{Q_Y(\tau \mid X)}$$

Esto le otorga al modelo la flexibilidad necesaria para capturar fenómenos complejos en los datos como:

$$\text{Asimetría} \qquad \text{Heterocedasticidad} \qquad \text{Comportamiento de colas}$$

de manera directa y sin requerir transformaciones previas de las variables.

# Comparativa de Bondad de Ajuste: $R^2$ vs. Pseudo-$R^2$

### 1. El $R^2$ en la Regresión Lineal (MCO)


* **Fórmula:**
  $$R^2 = 1 - \frac{\text{Suma de Cuadrados de los Residuos (SCR)}}{\text{Suma de Cuadrados Totales (SCT)}} = 1 - \frac{\sum (Y_i - \hat{Y}_i)^2}{\sum (Y_i - \bar{Y})^2}$$
* **¿Qué te dice del modelo?** Da una métrica global. Si obtenemos un $R^2 = 0.75$, significa que las variables explicativas logran capturar e interpretar el **75% de la varianza** o dispersión de los datos reales. Su escala va estrictamente de $0$ a $1$.

---

### 2. El Pseudo-$R^2$ ($R^1$) de Koenker y Machado en la Regresión Cuantílica

* **¿Qué mide?** La proporción de la reducción en la suma de pérdidas absolutas asimétricas (*check loss*) al pasar de un modelo vacío (solo el intercepto) al modelo completo para un cuantil específico.
* **Fórmula:**
  $$R^1(\tau) = 1 - \frac{\hat{V}(\tau)}{\tilde{V}(\tau)}$$
  *Donde $\hat{V}(\tau)$ es la pérdida total del modelo completo y $\tilde{V}(\tau)$ la del modelo nulo en el cuantil $\tau$.*
* **¿Qué  dice del modelo?** Da una métrica local. Te Indica qué tan bueno es un modelo explicando un sector específico de la distribución (las colas, la mediana, etc.). Por construcción matemática, sus valores **suelen ser notablemente más bajos** que el $R^2$ tradicional; un valor de $0.20$ ya suele considerarse un ajuste muy competitivo.

### Teorema 4.2: Distribución asintótica conjunta

**Teorema:** Sea una secuencia de cuantiles de regresión únicos $\beta^*(\theta_1), \dots, \beta^*(\theta_M)$ con $0 < \theta_1 < \dots < \theta_M < 1$. Bajo las condiciones de que la función de distribución $F$ tiene densidad continua y positiva $f$, y que la matriz de diseño converge a una matriz definida positiva $Q$ ($\lim_{T \to \infty} T^{-1} X'X = Q$), entonces:

$$
\sqrt{T}[\xi^*(\theta_1) - \xi(\theta_1), \dots, \xi^*(\theta_M) - \xi(\theta_M)] \xrightarrow{d} \mathcal{N}(0, \Omega \otimes Q^{-1})
$$

Donde $\Omega$ es la matriz de covarianzas de los $M$ cuantiles muestrales ordinarios de $F$.

---

### ¿Por qué se emplea este teorema? 

Este teorema es el **pilar teórico para realizar inferencia estadística** en regresión cuantílica para muestras grandes. Sus aplicaciones clave son:

* **Inferencia estadística:** Proporciona la distribución asintótica (normalidad) necesaria para calcular errores estándar, intervalos de confianza y valores $p$ de los coeficientes $\beta^*(\theta)$, ante la falta de una fórmula exacta para muestras finitas.
* **Comparación entre cuantiles:** Al definir la distribución conjunta de múltiples cuantiles mediante el producto de Kronecker ($\Omega \otimes Q^{-1}$), permite evaluar si el impacto de una variable difiere significativamente entre distintos puntos de la distribución (por ejemplo, comparar el percentil 10 contra el 90).
* **Extensión de la teoría clásica:** Demuestra de manera formal que los cuantiles de regresión preservan un comportamiento análogo al de los cuantiles muestrales tradicionales, integrando la estructura de las variables explicativas ($Q^{-1}$).
* **Sustento de robustez:** Permite demostrar matemáticamente que en distribuciones no normales o con valores atípicos, la regresión cuantílica puede ser más eficiente (menor varianza asintótica) que los Mínimos Cuadrados Ordinarios (MCO).

## Ejemplos, utilidad 


**Ejemplo 1: marketing.** El cuantil bajo ayuda a estudiar escenarios conservadores; el alto, capacidad potencial. La diferencia entre curvas describe heterogeneidad en la respuesta.

**Ejemplo 2: salarios.** La educación puede tener un efecto distinto en la cola baja y en la cola alta. Un único promedio puede ocultar desigualdad.

**Ejemplo 3: salud.** En peso al nacer, la misma covariable puede afectar de forma distinta a bebés con riesgo bajo y alto; esto motiva el análisis del cuaderno R.

**Ejemplo 4: rendimiento y riesgo.** En finanzas, cuantiles bajos modelan pérdidas extremas; en energía, cuantiles altos ayudan a planear demanda.
